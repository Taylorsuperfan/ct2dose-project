## N0 - Restore the independent code package

```python
from pathlib import Path
import base64, hashlib, io, zipfile, json, os
PAYLOAD_SHA256 = "8a39e6d44fc92502c2103290b35d8d385a28a17cecf2041d1d6c044200ed3513"
PAYLOAD_BASE64 = ""
payload = base64.b64decode(PAYLOAD_BASE64)
if hashlib.sha256(payload).hexdigest() != PAYLOAD_SHA256:
    raise RuntimeError("Notebook payload hash mismatch.")
CODE = Path("/content/conditional_wfr_dose_" + PAYLOAD_SHA256[:12])
CODE.mkdir(parents=True, exist_ok=True)
with zipfile.ZipFile(io.BytesIO(payload)) as archive:
    for name in archive.namelist():
        rel = Path(name)
        if rel.is_absolute() or ".." in rel.parts:
            raise RuntimeError("Unsafe archive entry.")
        destination = CODE / rel
        if not destination.resolve().is_relative_to(CODE.resolve()):
            raise RuntimeError("Archive destination escaped the code directory.")
        content = archive.read(name)
        if destination.exists() and destination.read_bytes() != content:
            raise RuntimeError("Existing restored code differs: " + str(destination))
        destination.parent.mkdir(parents=True, exist_ok=True)
        if not destination.exists():
            destination.write_bytes(content)
print("CODE:", CODE)
print("No data or trained checkpoint was opened.")
```

## N1 - Mount Drive and declare new output locations

```python
from google.colab import drive
from pathlib import Path
import subprocess, sys, os, signal, datetime, json, hashlib

drive.mount("/content/drive")
if not Path("/content/drive/MyDrive").is_dir():
    raise RuntimeError("Google Drive is not mounted.")

os.environ.update({
    "POT_BACKEND_DISABLE_JAX": "1",
    "POT_BACKEND_DISABLE_TENSORFLOW": "1",
    "POT_BACKEND_DISABLE_CUPY": "1",
    "XLA_PYTHON_CLIENT_PREALLOCATE": "false",
    "JAX_PLATFORMS": "cpu",
    "CWFR_CPU_THREADS": "2",
})
WORK = Path("/content/drive/MyDrive/ct2dose_thesis/conditional_wfr_dose")
CACHE = Path("/content/drive/MyDrive/ct2dose_thesis/phase9g_signed_real_v1/cache.local/train192_seed42")
PREVIOUS_HJD = Path("/content/drive/MyDrive/ct2dose_thesis/phase9g_signed_real_v1/evaluations.local/pilot_seed17_hjd_rf_eval01")
PREVIOUS_DIRECT = Path("/content/drive/MyDrive/ct2dose_thesis/phase9g_signed_real_v1/evaluations.local/pilot_seed17_residual_rf_eval01")
for name in ["code", "logs.local", "plans.local", "runs.local", "evaluations.local", "comparisons.local", "configs.local"]:
    (WORK / name).mkdir(parents=True, exist_ok=True)
backup = WORK / "code" / ("toolkit_" + PAYLOAD_SHA256[:12] + ".zip")
if backup.exists() and backup.read_bytes() != payload:
    raise RuntimeError("Saved source backup differs.")
if not backup.exists():
    backup.write_bytes(payload)
for stage in ["one", "six", "pilot"]:
    destination = WORK / "configs.local" / (stage + ".json")
    content = (CODE / "configs" / (stage + ".json")).read_bytes()
    if destination.exists() and destination.read_bytes() != content:
        raise RuntimeError("Existing configuration differs: " + str(destination))
    if not destination.exists():
        destination.write_bytes(content)

RUNS = {stage: WORK / "runs.local" / (stage + "_seed17") for stage in ["one", "six", "pilot"]}
PLANS = {stage: WORK / "plans.local" / stage for stage in ["one", "six", "pilot"]}
EVAL = WORK / "evaluations.local" / "pilot_seed17_val600"
COMPARE = WORK / "comparisons.local" / "pilot_seed17_vs_saved_baselines"
DEVICE = "cuda"
MAX_UPDATES_THIS_CALL = 128

def execute(arguments, cpu=False):
    stamp = datetime.datetime.now(datetime.timezone.utc).strftime("%Y%m%dT%H%M%S%fZ")
    env = os.environ.copy()
    env["PYTHONPATH"] = str(CODE)
    env["MPLBACKEND"] = "Agg"
    if cpu:
        env["CUDA_VISIBLE_DEVICES"] = ""
    command = [sys.executable, "-u", *map(str, arguments)]
    logpath = WORK / "logs.local" / (stamp + ".log")
    print("RUN:", " ".join(command), flush=True)
    with logpath.open("x", encoding="utf-8") as output:
        proc = subprocess.Popen(command, cwd=CODE, env=env, stdout=subprocess.PIPE,
            stderr=subprocess.STDOUT, text=True, bufsize=1, start_new_session=True)
        try:
            for line in proc.stdout:
                print(line, end="")
                output.write(line)
                output.flush()
            result = proc.wait()
        except KeyboardInterrupt:
            os.killpg(proc.pid, signal.SIGINT)
            try:
                proc.wait(timeout=15)
            except subprocess.TimeoutExpired:
                os.killpg(proc.pid, signal.SIGKILL)
                proc.wait()
            raise RuntimeError("Stopped. Resume from the last verified checkpoint only. Log: " + str(logpath))
    if result:
        raise RuntimeError("Command failed. Preserve the error and do not bypass checks. Log: " + str(logpath))
    print("LOG:", logpath)
    return logpath

def task(*arguments, cpu=False):
    return execute(["-m", "cwfr", *arguments], cpu=cpu)

print("NEW OUTPUT ROOT:", WORK)
print("READ-ONLY INPUT CACHE:", CACHE)
print("The previous medical arrays and predictions have not been opened.")
```
